```python
import re

def remove_toc_and_index(text: str) -> str:
    # normalise les fins de ligne
    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # supprime toute la table des matières
    text = re.sub(
        r"## Table of contents.*?(?=\n## Overview of compensation parameters\b)",
        "",
        text,
        flags=re.DOTALL
    )

    # supprime tout le keyword index
    text = re.sub(
        r"## Keyword index.*?(?=\n## 4 Appendix\b)",
        "",
        text,
        flags=re.DOTALL
    )

    # nettoie les lignes vides en trop
    text = re.sub(r"\n{3,}", "\n\n", text).strip()

    # supprimer les balises d'image 
    text = re.sub(r"!\[.*?\]\(.*?\)", "", text)

    return text
```

```python
import re
from typing import List

from transformers import AutoTokenizer
from langchain_core.documents import Document

from docling.document_converter import DocumentConverter
from docling_core.transforms.chunker.hybrid_chunker import HybridChunker
from docling_core.transforms.chunker.tokenizer.huggingface import HuggingFaceTokenizer
from docling_core.transforms.chunker.hierarchical_chunker import (
    ChunkingDocSerializer,
    ChunkingSerializerProvider,
)
from docling_core.transforms.serializer.markdown import MarkdownTableSerializer


FILE_PATH = "./../data/Manuals/mds_axis_compensation_en.pdf"
MODEL = "Qwen/Qwen3-Embedding-0.6B"


# ----------------------------
# 1) Serializer: tables -> Markdown
# ----------------------------
class MDTableSerializerProvider(ChunkingSerializerProvider):
    def get_serializer(self, doc):
        return ChunkingDocSerializer(
            doc=doc,
            table_serializer=MarkdownTableSerializer(),
        )


# ----------------------------
# 2) Nettoyage post-chunk
#    (on enlève TOC / Keyword index sans casser les tableaux)
# ----------------------------
def should_drop_chunk(text: str) -> bool:
    t = text.strip()

    # table des matières
    if "## Table of contents" in t or t.startswith("Table of contents"):
        return True

    # keyword index
    if "## Keyword index" in t or t.startswith("Keyword index"):
        return True

    # chunks de TOC très fréquents dans les manuels
    if re.search(r"P-COMP-\d{5}.*\[\}\s*\d+\]", t) and "Overview of compensation parameters" not in t:
        # à adapter selon ton doc
        return True

    return False


# ----------------------------
# 3) Conversion PDF -> DoclingDocument
# ----------------------------
converter = DocumentConverter()
conv_res = converter.convert(FILE_PATH)
dl_doc = conv_res.document


# ----------------------------
# 4) Tokenizer Docling aligné sur ton modèle d'embedding
# ----------------------------
base_tokenizer = AutoTokenizer.from_pretrained(MODEL, trust_remote_code=True)

hf_tokenizer = HuggingFaceTokenizer(
    tokenizer=base_tokenizer,
    max_tokens=500,
)


# ----------------------------
# 5) Chunker
#    - tables en Markdown
#    - répétition des headers de tableau
# ----------------------------
chunker = HybridChunker(
    tokenizer=hf_tokenizer,
    serializer_provider=MDTableSerializerProvider(),
    max_tokens=500,
    merge_peers=True,
    repeat_table_header=True,
    omit_header_on_overflow=True,
)


# ----------------------------
# 6) Création des chunks texte + metadata
# ----------------------------
raw_chunks = list(chunker.chunk(dl_doc=dl_doc))

docs: List[Document] = []

for i, chunk in enumerate(raw_chunks):
    text = chunker.contextualize(chunk=chunk).strip()

    if not text:
        continue

    if should_drop_chunk(text):
        continue

    meta = {
        "source": FILE_PATH,
        "chunk_id": i,
    }

    # si dispo, on garde les refs Docling
    if hasattr(chunk, "meta") and getattr(chunk, "meta", None) is not None:
        try:
            meta["doc_items"] = [it.self_ref for it in chunk.meta.doc_items]
        except Exception:
            pass

        try:
            meta["headings"] = chunk.meta.headings
        except Exception:
            pass

        try:
            meta["captions"] = chunk.meta.captions
        except Exception:
            pass

    docs.append(
        Document(
            page_content=text,
            metadata=meta
        )
    )


# ----------------------------
# 7) Debug
# ----------------------------
print(f"Nombre de chunks finaux: {len(docs)}\n")

for d in docs[:5]:
    print("=" * 120)
    print(d.page_content[:2000])
    print("\nMETA:", d.metadata)
```
